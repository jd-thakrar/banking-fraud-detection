# Banking Fraud Detection — Data Understanding

This notebook performs the initial understanding and quality assessment of
the banking transaction dataset.

The objectives are:

- Load the original dataset
- Understand its structure and dimensions
- Inspect data types
- Identify missing values
- Identify duplicate records
- Examine categorical and numerical features
- Analyze the fraud class distribution
- Identify potential data-quality issues

No data transformation or machine learning model is performed in this notebook.
